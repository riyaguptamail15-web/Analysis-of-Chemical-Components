```python
# Keep only moisturizers
moisturizers = df[df["Label"] == "Moisturizer"]

# View first 5 rows
moisturizers.head()
```

```python
# Keep moisturizers suitable for dry skin
moisturizers_dry = moisturizers[moisturizers["Dry"] == 1]

# Reset index
moisturizers_dry = moisturizers_dry.reset_index(drop=True)

# Check size
moisturizers_dry
```

## TASK 3

```python
# Create a working copy
moisturizers = moisturizers_dry.copy()
```

```python
# Convert ingredient text to lowercase as Machine learning treats different cases as different words, so we standardize everything.
moisturizers["Ingredients"] = moisturizers["Ingredients"].str.lower()
```

```python
# Split ingredients into a list
moisturizers["Tokens"] = moisturizers["Ingredients"].str.split(", ")
```

```python
moisturizers["Tokens"].iloc[0]
```

```python
# #Python creates:

# Ingredient
# Index
# water 0
# glycerin 1
# aloe 2
# vitamin e 3

# This dictionary is called ingredient_idx.
# It maps every ingredient to a unique number.
    
ingredient_idx = {}
idx = 0

for ingredients in moisturizers["Tokens"]:
    for ingredient in ingredients:
        if ingredient not in ingredient_idx:
            ingredient_idx[ingredient] = idx
            idx += 1
```

```python
print("Total unique ingredients:", len(ingredient_idx))
```

```python
list(ingredient_idx.items())[:30]
```

## TASK 4 

```python
# Number of products
M = len(moisturizers)

# Number of unique ingredients
N = len(ingredient_idx)

print("Products:", M)
print("Ingredients:", N)
```

```python
# Create empty Document-Term Matrix
A = np.zeros((M, N))
```

```python
A.shape
```

```python
A
```

## TASK 5

```python
ingredient_idx
```

```python
def oh_encoder(tokens):

    # Create empty vector
    x = np.zeros(N)

    # Fill 1 where ingredient exists
    for ingredient in tokens:
        if ingredient in ingredient_idx:
            x[ingredient_idx[ingredient]] = 1

    return x
```

```python
sample_tokens = moisturizers["Tokens"].iloc[78]

encoded = oh_encoder(sample_tokens)

print(encoded[:1000])
```

```python
print("Ingredients in this product:", int(encoded.sum()))
```

## TASK 6

```python
# Fill the Document-Term Matrix
# enumerate() gives you both the row number (i) and the ingredient list (tokens).
# oh_encoder(tokens) converts ingredients into 1s and 0s.
# A[i] stores that binary vector in the matrix.

for i, tokens in enumerate(moisturizers["Tokens"]):
    A[i] = oh_encoder(tokens)

print("Matrix generated successfully!")
```

## TASK 7

What is t-SNE?

Right now every moisturizer has 2233 features.
Humans can't visualize 2233 dimensions.
t-SNE compresses them into just 2 coordinates while keeping similar products close together.

```python
#Import TSNE
from sklearn.manifold import TSNE
```

```python
tsne = TSNE(
    n_components=2,
    learning_rate=200,
    random_state=42,
    init="random"
)
```

```python
tsne_result = tsne.fit_transform(A)
```

```python
moisturizers["X"] = tsne_result[:, 0]
moisturizers["Y"] = tsne_result[:, 1]

moisturizers[["Brand","Name","Price","Rank", "X","Y"]].head()
```

```python
import re

# Clean and tokenize ingredients
moisturizers["Tokens"] = moisturizers["Ingredients"].apply(
    lambda text: [
        re.sub(r'[\*\.]+$', '', ing.strip().lower())   # remove trailing *, ** and .
        for ing in text.split(",")
        if ing.strip()
    ]
)
```

```python
product1 = moisturizers.iloc[32]
product2 = moisturizers.iloc[80]
product3 = moisturizers.iloc[75]

print("PRODUCT 1")
print(product1["Brand"])
print(product1["Name"])

print("\nPRODUCT 2")
print(product2["Brand"])
print(product2["Name"])

print("\nPRODUCT 3")
print(product2["Brand"])
print(product2["Name"])
```

```python
print("Ingredients of Product 1:\n")
print(product1["Ingredients"])

print("\n" + "="*80 + "\n")

print("Ingredients of Product 2:\n")
print(product2["Ingredients"])

print("\n" + "="*80 + "\n")

print("Ingredients of Product 3:\n")
print(product3["Ingredients"])
```

```python
# ---------- Similarity Comparison (Products 32, 75, 80) ----------

pairs = [(32, 75), (32, 80), (75, 80)]

for idx1, idx2 in pairs:
    set1 = set(moisturizers.loc[idx1, "Tokens"])
    set2 = set(moisturizers.loc[idx2, "Tokens"])

    common = sorted(set1.intersection(set2))

    print("="*90)
    print(f"Product {idx1}: {moisturizers.loc[idx1, 'Brand']} - {moisturizers.loc[idx1, 'Name']}")
    print(f"Product {idx2}: {moisturizers.loc[idx2, 'Brand']} - {moisturizers.loc[idx2, 'Name']}")
    print(f"\nNumber of Common Ingredients: {len(common)}\n")

    for ingredient in common:
        print("-", ingredient)

    print("\n")
```

## Conclusion 1: Similar Products (Products 32, 75 & 80)

Conclusion: The comparison of Products 32, 75, and 80 showed varying levels of ingredient similarity. Products 75 and 80 shared the highest number of common ingredients, indicating a closely related moisturizing formulation. Product 32 shared comparatively fewer ingredients with the other two, suggesting it belongs to a different botanical oil-based formulation. This supports the t-SNE visualization, where products positioned closer together exhibit greater chemical similarity.

### Note:
Products 32 and 75 showed 0 common ingredients and Products 32 and 80 showed only 1 common ingredient in the token comparison. This occurs because the project uses basic tokenization, where ingredients are matched as exact strings. Minor differences in formatting (extra spaces, */** symbols, singular vs. plural words, or slight naming variations such as caprylic/capric triglycerides vs. caprylic/ capric triglyceride) are treated as different tokens. Therefore, the actual chemical similarity is higher than the reported overlap, and advanced text normalization would be required to identify these equivalent ingredients.
